## 1. Basic Data Science Workflow

A typical data science workflow follows these steps:
1. Data Collection
2. Data Cleaning
3. Exploratory Data Analysis
4. Modeling
5. Interpretation & Communication

Let's see how this works with a simple example.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot styling
plt.figure(figsize=(10, 6))
sns.set(style="whitegrid")
```

### Step 1: Data Collection
In a real scenario, you might collect data from databases, APIs, or files. Here, we'll create a simple dataset.

```python
# Create a sample dataset of student performance
data = {
    'student_id': range(1, 11),
    'hours_studied': [2, 5, 3, 8, 6, 9, 4, 7, 6, 5],
    'previous_score': [65, 78, 72, 90, 83, 94, 75, 85, 80, 76],
    'exam_score': [61, 82, 70, 95, 87, 98, 73, 89, 85, 78]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display the first few rows
df.head()
```

### Step 2: Exploratory Data Analysis (EDA)
Let's explore the data to understand patterns and relationships.

```python
# Basic statistics
print("Basic statistics:")
print(df.describe())

# Correlation between variables
print("\nCorrelation matrix:")
print(df.corr())

# Visualize the correlation between hours studied and exam score
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='hours_studied', y='exam_score')
plt.title("Hours Studied vs. Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.show()
```

### Step 3: Data Modeling
Let's build a simple linear regression model to predict exam scores.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Prepare features (X) and target (y)
X = df[['hours_studied', 'previous_score']]
y = df['exam_score']

# Create and fit the model
model = LinearRegression()
model.fit(X, y)

# Make predictions
y_pred = model.predict(X)

# Print model coefficients
print(f"Intercept: {model.intercept_:.2f}")
print(f"Coefficient for hours studied: {model.coef_[0]:.2f}")
print(f"Coefficient for previous score: {model.coef_[1]:.2f}")

# Evaluate the model
print(f"R² Score: {r2_score(y, y_pred):.2f}")
print(f"Mean Squared Error: {mean_squared_error(y, y_pred):.2f}")
```

### Step 4: Interpretation
Let's interpret our model and make predictions.

```python
# Create a comparison of actual vs predicted values
results = pd.DataFrame({'Actual': y, 'Predicted': y_pred})
print(results)

# Make a prediction for a new student
new_student = pd.DataFrame({
    'hours_studied': [7],
    'previous_score': [88]
})

predicted_score = model.predict(new_student)
print(f"\nPredicted score for a student who studied 7 hours and had a previous score of 88: {predicted_score[0]:.2f}")

# What this means
print("\nInterpretation:")
print(f"- For each additional hour studied, the exam score is expected to increase by {model.coef_[0]:.2f} points")
print(f"- For each point increase in previous score, the exam score is expected to increase by {model.coef_[1]:.2f} points")
```

## 2. Introduction to AI Concepts

Let's demonstrate a simple machine learning classification task with the Iris dataset.

```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Load the iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# Split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train a Random Forest model (an example of machine learning)
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

# Make predictions
y_pred = rf.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Model accuracy: {accuracy:.2f}")

# Example: Predicting the species of a new flower
# Features: sepal length, sepal width, petal length, petal width
new_flower = [[5.1, 3.5, 1.4, 0.2]]  # Example measurements
prediction = rf.predict(new_flower)
print(f"\nPredicted species for new flower: {iris.target_names[prediction[0]]}")
```

## 3. Ethical Considerations in Data Science

Here's an example showing the importance of considering ethics in data science:

```python
# Create a biased dataset for a fictional loan approval system
np.random.seed(42)

# Sample size
n = 1000

# Create features
age = np.random.normal(35, 10, n).astype(int)  # Age between ~15-55
income = np.random.normal(50000, 15000, n).astype(int)  # Income around $50k
debt = np.random.normal(15000, 10000, n).astype(int)  # Debt around $15k
gender = np.random.choice(['Male', 'Female'], n)  # Gender

# Create a biased dataset where females are less likely to get approved despite similar profiles
approval_probability = np.where(
    gender == 'Male',
    0.7 + 0.2 * (income > 50000) - 0.3 * (debt > 20000),  # Males: 70% baseline, +20% for high income, -30% for high debt
    0.5 + 0.2 * (income > 50000) - 0.3 * (debt > 20000)   # Females: 50% baseline (biased), other factors same
)

# Generate approval outcomes
approved = np.random.binomial(1, approval_probability)

# Create DataFrame
loan_data = pd.DataFrame({
    'Age': age, 
    'Income': income,
    'Debt': debt,
    'Gender': gender,
    'Approved': approved
})

# Display the data
print(loan_data.head())

# Analyze approval rates by gender
approval_by_gender = loan_data.groupby('Gender')['Approved'].mean()
print("\nApproval Rates by Gender:")
print(approval_by_gender)

# Visualize the bias
plt.figure(figsize=(10, 6))
sns.barplot(x='Gender', y='Approved', data=loan_data)
plt.title('Loan Approval Rates by Gender')
plt.ylabel('Approval Rate')
plt.ylim(0, 1)
plt.show()

# Compare approval rates for similar income levels
loan_data['Income_Level'] = pd.qcut(loan_data['Income'], 4, labels=['Low', 'Medium-Low', 'Medium-High', 'High'])

plt.figure(figsize=(12, 6))
sns.barplot(x='Income_Level', y='Approved', hue='Gender', data=loan_data)
plt.title('Loan Approval Rates by Income Level and Gender')
plt.ylabel('Approval Rate')
plt.ylim(0, 1)
plt.show()

print("\nThis example demonstrates how an AI system trained on biased historical data")
print("would perpetuate and potentially amplify that bias in its decisions.")
print("Data scientists must carefully examine their training data for such biases")
print("and implement fairness constraints to ensure ethical AI systems.")
```
