## 2. Data Structures for Data Science

```python
# Lists - ordered, mutable collections
languages = ["Python", "R", "SQL", "Julia"]
print(f"Languages list: {languages}")
print(f"First language: {languages[0]}")
print(f"Last language: {languages[-1]}")

# Adding elements
languages.append("Scala")
print(f"After append: {languages}")

# List slicing
print(f"First two languages: {languages[:2]}")
print(f"Last two languages: {languages[-2:]}")

# List comprehension (powerful for data transformation)
lengths = [len(lang) for lang in languages]
print(f"Character counts: {lengths}")

# Filtering with list comprehension
long_names = [lang for lang in languages if len(lang) > 4]
print(f"Languages with >4 letters: {long_names}")
```

```python
# Dictionaries - key-value pairs (similar to JSON data)
data_scientist = {
    "name": "Alex",
    "skills": ["Python", "SQL", "Machine Learning"],
    "experience_years": 5,
    "education": {"degree": "Masters", "field": "Data Science"}
}

# Accessing dictionary values
print(f"Name: {data_scientist['name']}")
print(f"First skill: {data_scientist['skills'][0]}")
print(f"Education field: {data_scientist['education']['field']}")

# Adding new key-value pairs
data_scientist["location"] = "Remote"
print(f"Updated dictionary: {data_scientist}")

# Dictionary methods
print(f"Keys: {data_scientist.keys()}")
print(f"Values: {list(data_scientist.values())}")
print(f"Items: {list(data_scientist.items())[:2]}...")

# Safe key access with get()
print(f"Salary: {data_scientist.get('salary', 'Not specified')}")
```

```python
# Sets - unordered collections of unique elements (great for deduplication)
user_ids = {101, 102, 103, 101, 104, 102, 105}
print(f"Unique user IDs: {user_ids}")

# Set operations
group_a = {101, 102, 103}
group_b = {103, 104, 105}

print(f"Union: {group_a | group_b}")  # All users from both groups
print(f"Intersection: {group_a & group_b}")  # Users in both groups
print(f"Difference A-B: {group_a - group_b}")  # Users only in group A
print(f"Symmetric difference: {group_a ^ group_b}")  # Users in exactly one group
```

```python
# Tuples - immutable sequences (good for fixed data)
coordinates = (40.7128, -74.0060)  # New York coordinates
print(f"Latitude: {coordinates[0]}, Longitude: {coordinates[1]}")

# Tuple unpacking (common in data science)
latitude, longitude = coordinates
print(f"Unpacked - Lat: {latitude}, Long: {longitude}")

# Named tuples (even better for readability)
from collections import namedtuple
Location = namedtuple('Location', ['lat', 'long'])
ny = Location(lat=40.7128, long=-74.0060)
print(f"Named tuple - Lat: {ny.lat}, Long: {ny.long}")
```

## 3. Control Flow and Functions

```python
# Conditional statements
def classify_data_size(num_records):
    if num_records < 1000:
        return "Small dataset"
    elif num_records < 100000:
        return "Medium dataset"
    elif num_records < 10000000:
        return "Large dataset"
    else:
        return "Big data"

print(classify_data_size(500))
print(classify_data_size(50000))
print(classify_data_size(5000000))
print(classify_data_size(50000000))
```

```python
# Loops - for processing datasets
sample_data = [12, 15, 23, 8, 42, 16, 4, 8]

# For loop with aggregation
total = 0
for value in sample_data:
    total += value
print(f"Sum: {total}, Mean: {total/len(sample_data):.2f}")

# While loop example - processing until condition met
def find_threshold_index(data, threshold):
    index = 0
    cumulative = 0
    while index < len(data) and cumulative < threshold:
        cumulative += data[index]
        index += 1
    return index

print(f"Index where sum exceeds 50: {find_threshold_index(sample_data, 50)}")

# Enumerate - when you need both index and value
for i, value in enumerate(sample_data):
    if value > 20:
        print(f"Value {value} at position {i} exceeds threshold")
```

```python
# Functions with default parameters
def normalize_data(values, min_val=None, max_val=None):
    """Normalize data to range [0,1].
    
    Parameters:
    values - list of numeric values
    min_val - minimum value for scaling (if None, use min of values)
    max_val - maximum value for scaling (if None, use max of values)
    
    Returns:
    list of normalized values
    """
    if min_val is None:
        min_val = min(values)
    if max_val is None:
        max_val = max(values)
        
    range_val = max_val - min_val
    if range_val == 0:
        return [0.5] * len(values)  # Avoid division by zero
        
    return [(x - min_val) / range_val for x in values]

data = [10, 20, 30, 40, 50]
print(f"Original data: {data}")
print(f"Normalized data: {normalize_data(data)}")
print(f"Custom range normalized: {normalize_data(data, 0, 100)}")
```

```python
# Args and kwargs - flexible parameter handling

def calculate_metrics(*values, **options):
    """Calculate statistical metrics for a set of values.
    
    Parameters:
    *values - variable number of data points
    **options - configuration options (e.g. precision, include_std)
    """
    if not values:
        return None
        
    count = len(values)
    total = sum(values)
    mean = total / count
    
    results = {
        "count": count,
        "sum": total,
        "mean": mean
    }
    
    # Calculate additional metrics based on options
    if options.get("include_std", False):
        import math
        variance = sum((x - mean) ** 2 for x in values) / count
        std_dev = math.sqrt(variance)
        results["std_dev"] = std_dev
    
    # Apply precision formatting
    precision = options.get("precision", 2)
    for key in results:
        if isinstance(results[key], float):
            results[key] = round(results[key], precision)
    
    return results

# Example usage
print(calculate_metrics(12, 15, 23, 8, 42, 16))
print(calculate_metrics(12, 15, 23, 8, 42, 16, include_std=True, precision=4))
```

## 4. Object-Oriented Programming for Data Science

```python
class Dataset:
    """A class to represent and analyze a dataset."""
    
    def __init__(self, name, data=None):
        self.name = name
        self.data = data if data is not None else []
        self.metadata = {}
    
    def add_record(self, record):
        """Add a single record to the dataset."""
        self.data.append(record)
    
    def add_records(self, records):
        """Add multiple records to the dataset."""
        self.data.extend(records)
    
    def set_metadata(self, key, value):
        """Add metadata about the dataset."""
        self.metadata[key] = value
    
    def describe(self):
        """Get basic statistics about the dataset."""
        if not self.data:
            return "Empty dataset"
        
        # Only works for numeric datasets
        if all(isinstance(x, (int, float)) for x in self.data):
            return {
                "count": len(self.data),
                "min": min(self.data),
                "max": max(self.data),
                "mean": sum(self.data) / len(self.data)
            }
        else:
            return {
                "count": len(self.data),
                "types": set(type(x).__name__ for x in self.data)
            }
    
    def __str__(self):
        return f"Dataset '{self.name}' with {len(self.data)} records"
    
    def __len__(self):
        return len(self.data)

# Demo of the Dataset class
temperature_data = Dataset("Daily Temperatures")
temperature_data.add_records([68, 71, 72, 69, 75, 73, 70])
temperature_data.set_metadata("unit", "Fahrenheit")
temperature_data.set_metadata("location", "New York")

print(temperature_data)
print(f"Metadata: {temperature_data.metadata}")
print(f"Statistics: {temperature_data.describe()}")
```

```python
# Inheritance example - specialized dataset
class TimeSeriesDataset(Dataset):
    """A dataset with time-indexed data points."""
    
    def __init__(self, name, timestamps=None, values=None):
        super().__init__(name)
        self.timestamps = timestamps if timestamps is not None else []
        self.values = values if values is not None else []
        
        # Override data property to use timestamp-value pairs
        if len(self.timestamps) == len(self.values):
            self.data = list(zip(self.timestamps, self.values))
    
    def add_point(self, timestamp, value):
        """Add a time-value pair."""
        self.timestamps.append(timestamp)
        self.values.append(value)
        self.data.append((timestamp, value))
    
    def get_values_in_range(self, start_time, end_time):
        """Get values within a time range."""
        return [(t, v) for t, v in self.data if start_time <= t <= end_time]
    
    def describe(self):
        """Enhanced statistics for time series data."""
        basic_stats = super().describe()
        if isinstance(basic_stats, str):
            return basic_stats
            
        if self.timestamps:
            basic_stats["time_range"] = (min(self.timestamps), max(self.timestamps))
        
        if self.values and all(isinstance(x, (int, float)) for x in self.values):
            basic_stats["value_stats"] = {
                "min": min(self.values),
                "max": max(self.values),
                "mean": sum(self.values) / len(self.values)
            }
            
        return basic_stats

# Demo of the TimeSeriesDataset
import datetime

# Create timestamps for a week
base_date = datetime.datetime(2025, 4, 1)
dates = [base_date + datetime.timedelta(days=i) for i in range(7)]
stock_prices = [145.8, 146.2, 147.5, 145.9, 148.2, 149.7, 150.3]

stock_data = TimeSeriesDataset("AAPL Stock Prices", dates, stock_prices)
stock_data.set_metadata("currency", "USD")

print(stock_data)
print(f"First 3 data points: {stock_data.data[:3]}")
print(f"Statistics: {stock_data.describe()}")

# Get prices for a specific range
start = datetime.datetime(2025, 4, 3)
end = datetime.datetime(2025, 4, 5)
selected_range = stock_data.get_values_in_range(start, end)
print(f"Data from {start.date()} to {end.date()}: {selected_range}")
```

## 5. Working with Files and Error Handling

```python
# File handling with context managers
import os

# Write some sample data
sample_data = """date,temperature,humidity,wind_speed
2025-04-01,72,65,5
2025-04-02,75,68,7
2025-04-03,71,70,8
2025-04-04,69,72,10
2025-04-05,73,67,6
"""

filename = "weather_data.csv"

# Write data to file
try:
    with open(filename, 'w') as file:
        file.write(sample_data)
    print(f"Data written to {filename}")
except IOError as e:
    print(f"Error writing file: {e}")

# Read and parse the data
try:
    weather_data = []
    with open(filename, 'r') as file:
        # Skip header row
        header = next(file).strip().split(',')
        for line in file:
            values = line.strip().split(',')
            record = {}
            for i, field in enumerate(header):
                # Convert numeric values
                if i > 0:  # Skip date field
                    record[field] = int(values[i])
                else:
                    record[field] = values[i]
            weather_data.append(record)
    
    print(f"Read {len(weather_data)} records:")
    for record in weather_data:
        print(f"  {record['date']}: {record['temperature']}°F, {record['humidity']}% humidity")
        
except FileNotFoundError:
    print(f"File {filename} not found")
except Exception as e:
    print(f"Error processing file: {e}")
finally:
    # Clean up the file
    if os.path.exists(filename):
        os.remove(filename)
        print(f"Removed temporary file {filename}")
```

## 6. Advanced Python Features for Data Science

```python
# Lambda functions (anonymous functions)
data = [15, 8, 22, 43, 12, 7, 18]

# Sort by absolute difference from the mean
mean = sum(data) / len(data)
print(f"Original data: {data}")
print(f"Mean: {mean:.2f}")

sorted_by_distance = sorted(data, key=lambda x: abs(x - mean))
print(f"Sorted by distance from mean: {sorted_by_distance}")

# Find values above threshold using filter
threshold = 15
above_threshold = list(filter(lambda x: x > threshold, data))
print(f"Values above {threshold}: {above_threshold}")

# Transform values using map
normalized = list(map(lambda x: (x - min(data)) / (max(data) - min(data)), data))
print(f"Normalized values: {[round(x, 2) for x in normalized]}")
```

```python
# Generators and generator expressions
def moving_average(data, window_size):
    """Generate moving averages from data stream."""
    window = []
    for value in data:
        window.append(value)
        if len(window) > window_size:
            window.pop(0)
        if len(window) == window_size:
            yield sum(window) / window_size

# Generate some test data
sensor_data = [12, 14, 16, 13, 15, 17, 19, 21, 18, 16, 15, 17]
print(f"Sensor data: {sensor_data}")

# Calculate moving averages with window size 3
averages = list(moving_average(sensor_data, 3))
print(f"Moving averages (window=3): {[round(x, 2) for x in averages]}")

# Generator expressions - compact way to create generators
positive_values = (x for x in sensor_data if x > 15)
print(f"Values > 15: {list(positive_values)}")

# Calculate squares of values
squared_values = (x**2 for x in range(1, 11))
print(f"Squares from 1 to 10: {list(squared_values)}")
```

```python
# Decorators - useful for timing, logging, caching in data pipelines
import time
from functools import wraps

def timing_decorator(func):
    """A decorator that reports the execution time."""
    
    @wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"{func.__name__} executed in {(end_time - start_time):.6f} seconds")
        return result
    
    return wrapper

# Apply the decorator to a function
@timing_decorator
def compute_factorials(n):
    """Calculate factorials from 1 to n."""
    results = []
    for i in range(1, n+1):
        factorial = 1
        for j in range(1, i+1):
            factorial *= j
        results.append(factorial)
    return results

# Test the decorated function
factorials = compute_factorials(20)
print(f"Computed {len(factorials)} factorials")
print(f"First 5 factorials: {factorials[:5]}")
print(f"Last factorial: {factorials[-1]}")
```
