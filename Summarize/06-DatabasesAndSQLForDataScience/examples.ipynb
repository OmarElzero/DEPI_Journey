## 4. Filtering Techniques

SQL offers various ways to filter data:

```python
%%sql
-- Comparison operators
SELECT * FROM Employees WHERE Age >= 30 AND Age <= 40;
```

```python
%%sql
-- BETWEEN operator (equivalent to the above query)
SELECT * FROM Employees WHERE Age BETWEEN 30 AND 40;
```

```python
%%sql
-- IN operator for multiple values
SELECT * FROM Employees WHERE DepartmentID IN (1, 3);
```

```python
%%sql
-- LIKE operator for pattern matching
SELECT * FROM Employees WHERE LastName LIKE 'S%';
```

```python
%%sql
-- Handling NULL values
SELECT * FROM Employees WHERE DepartmentID IS NULL;
```

## 5. Aggregation Functions

SQL provides built-in functions for data analysis:

```python
%%sql
-- Count employees
SELECT COUNT(*) AS TotalEmployees FROM Employees;
```

```python
%%sql
-- Multiple aggregate functions
SELECT 
    COUNT(*) AS TotalEmployees,
    AVG(Age) AS AverageAge,
    MIN(Age) AS YoungestAge,
    MAX(Age) AS OldestAge
FROM Employees;
```

```python
%%sql
-- Group data with GROUP BY
SELECT 
    Gender, 
    COUNT(*) AS Count,
    ROUND(AVG(Age), 1) AS AverageAge
FROM Employees
GROUP BY Gender;
```

```python
%%sql
-- Filter groups with HAVING
SELECT 
    DepartmentID,
    COUNT(*) AS EmployeeCount,
    ROUND(AVG(Age), 1) AS AverageAge
FROM Employees
GROUP BY DepartmentID
HAVING COUNT(*) > 1;
```

## 6. Joining Tables

Joining tables is essential for working with relational databases:

```python
%%sql
-- INNER JOIN: Only returns matching records
SELECT 
    e.EmployeeID,
    e.FirstName,
    e.LastName,
    d.DepartmentName
FROM 
    Employees e
INNER JOIN 
    Departments d ON e.DepartmentID = d.DepartmentID;
```

```python
%%sql
-- LEFT JOIN: Keeps all records from the left table
SELECT 
    e.EmployeeID,
    e.FirstName,
    e.LastName,
    d.DepartmentName
FROM 
    Employees e
LEFT JOIN 
    Departments d ON e.DepartmentID = d.DepartmentID;
```

```python
%%sql
-- Multiple table joins
SELECT 
    e.FirstName,
    e.LastName,
    d.DepartmentName,
    s.Salary
FROM 
    Employees e
JOIN 
    Departments d ON e.DepartmentID = d.DepartmentID
JOIN 
    Salaries s ON e.EmployeeID = s.EmployeeID
ORDER BY 
    s.Salary DESC;
```

## 7. Advanced SQL Techniques

```python
%%sql
-- Subquery example
SELECT 
    FirstName, 
    LastName,
    Age
FROM 
    Employees
WHERE 
    Age > (SELECT AVG(Age) FROM Employees);
```

```python
%%sql
-- CASE statement for conditional logic
SELECT 
    FirstName,
    LastName,
    Age,
    CASE 
        WHEN Age < 30 THEN 'Young'
        WHEN Age BETWEEN 30 AND 40 THEN 'Mid-age'
        ELSE 'Senior'
    END AS AgeCategory
FROM 
    Employees
ORDER BY 
    Age;
```

```python
%%sql
-- Common Table Expression (CTE)
WITH DepartmentStats AS (
    SELECT 
        DepartmentID,
        COUNT(*) AS EmpCount,
        AVG(Age) AS AvgAge
    FROM 
        Employees
    GROUP BY 
        DepartmentID
)
SELECT 
    d.DepartmentName,
    ds.EmpCount,
    ROUND(ds.AvgAge, 1) AS AverageAge
FROM 
    DepartmentStats ds
JOIN 
    Departments d ON ds.DepartmentID = d.DepartmentID
ORDER BY 
    ds.EmpCount DESC;
```

## 8. SQL for Data Analysis

Using SQL for typical data analysis tasks:

```python
%%sql
-- Window functions
SELECT 
    e.FirstName,
    e.LastName,
    d.DepartmentName,
    s.Salary,
    AVG(s.Salary) OVER (PARTITION BY e.DepartmentID) AS AvgDeptSalary,
    s.Salary - AVG(s.Salary) OVER (PARTITION BY e.DepartmentID) AS DiffFromAvg
FROM 
    Employees e
JOIN 
    Departments d ON e.DepartmentID = d.DepartmentID
JOIN 
    Salaries s ON e.EmployeeID = s.EmployeeID
ORDER BY 
    d.DepartmentName, s.Salary DESC;
```

```python
%%sql
-- Ranking functions
SELECT 
    e.FirstName,
    e.LastName,
    d.DepartmentName,
    s.Salary,
    RANK() OVER (PARTITION BY e.DepartmentID ORDER BY s.Salary DESC) AS SalaryRank
FROM 
    Employees e
JOIN 
    Departments d ON e.DepartmentID = d.DepartmentID
JOIN 
    Salaries s ON e.EmployeeID = s.EmployeeID
ORDER BY 
    d.DepartmentName, SalaryRank;
```

## 9. Integration with Pandas for Data Analysis

Combining SQL with pandas for data analysis:

```python
# SQL query results to pandas DataFrame
import pandas as pd
import sqlite3

conn = sqlite3.connect('example.db')

query = """
SELECT 
    e.FirstName,
    e.LastName,
    e.Age,
    e.Gender,
    d.DepartmentName,
    s.Salary
FROM 
    Employees e
LEFT JOIN 
    Departments d ON e.DepartmentID = d.DepartmentID
LEFT JOIN 
    Salaries s ON e.EmployeeID = s.EmployeeID
"""

df = pd.read_sql(query, conn)
df
```

```python
# Further analysis with pandas
df.describe()
```

```python
# Analysis by department
department_analysis = df.groupby('DepartmentName').agg({
    'Age': ['mean', 'min', 'max'],
    'Salary': ['mean', 'min', 'max', 'count']
})
department_analysis
```

```python
# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the figure size
plt.figure(figsize=(10, 6))

# Create the salary distribution by department
sns.boxplot(x='DepartmentName', y='Salary', data=df)
plt.title('Salary Distribution by Department')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
